### ✅ 예상 출력
```
절차적: PYTHON IS FUN
객체지향: PYTHON IS FUN
```

### 💡 힌트
- `ord('a') == 97`, `ord('A') == 65` → 차이는 32
- 소문자 판단: `'a' <= ch <= 'z'`

````python
text = "python is fun"

# (A) 절차적 방식
result_a = ""
for ch in text:
    # TODO 3-1: 소문자면 대문자로 변환해서 result_a에 누적,
    #          아니면 그대로 추가
    if 'a' <= ch <= 'z':
        result_a += chr(ord(ch) - 32)
    else:
        result_a += ch
print("절차적:", result_a)

# (B) 객체 메소드 방식
# TODO 3-2: text.upper() 사용
result_b = text.upper()
print(result_b)

# 문자열은 파이썬에서 immutable이다.
# 한번 메모리에 만들어지면 원본은 누가 와도 절대 쪼개거나 바꿀 수 없다.

# 리스트는 마음대로 주무를 수 있는 찰흙과 같은 mutable 가변성이다.
# 원본을 뜯어고칠 수 있어서 번거롭게 복사본을 만들어서 반환할 필요가 없다.

# 
````

## 🧪 Lab 5. 생성자 `__init__`

### 📋 과제
Lab 4의 `Book` 클래스에 `__init__`을 추가하세요.
- 매개변수: `title`, `author`, `pages`
- `describe()`는 `[책] 제목: ○○, 저자: △△, 쪽수: ☐☐쪽` 형식으로 출력

### ✅ 예상 출력
```
[책] 제목: 어린 왕자, 저자: 생텍쥐페리, 쪽수: 120쪽
[책] 제목: 파이썬 입문, 저자: 김교수, 쪽수: 350쪽
[책] 제목: 데이터 과학, 저자: 박연구원, 쪽수: 480쪽
```

````python
# __init__은 영어 Initialize 의 양옆에 던더를 붙인 특별한 메서드이다.
# 인스턴스를 막 구워냈을 때, 태어나자마자 초기 정보(출생신고서)를 딱 박아주는 역할을 한다.

# self.title = title은 왜 두 번이나 쓸까?
# 파이썬 객체 지향의 핵심이다. 좌우의 역할이 완전히 다르다.
# 우측의 title은 임시재료이다.
# self.title : 영구적인 이름표이다.
# 밖에서 던져준 임시재료를 받아서 내 가슴팍의 고정 이름표에 매직으로 눌러 적겠다 라는 뜻이다.
# 이렇게 적어두면 나중에 다른 메서드에서도 언제든 self.title을 꺼내볼 수 있다.



class Book:
    def __init__(self, title, author, pages):
        # TODO 5-1: self.title, self.author, self.pages 초기화
        self.title = title
        self.author = author
        self.pages = pages
        
    def describe(self):
        # TODO 5-2: 위 형식대로 출력
        print(f'[책] 제목: {self.title}, 저자: {self.author}, 쪽수: {self.pages}')


book1 = Book("어린 왕자", "생텍쥐페리", 120)
book2 = Book("파이썬 입문", "김교수", 350)
book3 = Book("데이터 과학", "박연구원", 480)

# TODO 5-3: 세 인스턴스 모두 describe() 호출
book1.describe()
book2.describe()
book3.describe()
````

## 🧪 Lab 7. 캡슐화와 setter/getter

### 📋 과제
`BankAccount` 클래스를 만드세요.
- 속성: `owner`, `balance`
- 메소드:
  - `deposit(amount)`: 양수일 때만 입금, 아니면 `"잘못된 금액입니다."`
  - `withdraw(amount)`: 양수이고 잔액보다 작거나 같을 때만 출금
    - 음수/0이면 `"잘못된 금액입니다."`
    - 잔액 부족이면 `"잔액 부족입니다."`
  - `get_balance()`: 잔액 반환

### ✅ 예상 출력
```
입금 후 잔액: 15000
잘못된 금액입니다.
잔액: 15000
잔액 부족입니다.
잔액: 15000
출금 후 잔액: 8000
```

````python
class BankAccount:
    def __init__(self, owner, balance):
        # TODO 7-1: self.owner, self.balance 초기화
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        # TODO 7-2: amount > 0 이면 잔액 증가, 아니면 메시지 출력
        if amount > 0:
            self.balance += amount
        else:
            print('잘못된 금액입니다.')
    
    def withdraw(self, amount):
        # TODO 7-3: 조건 분기
        # - amount <= 0: "잘못된 금액입니다."
        # - amount > self.balance: "잔액 부족입니다."
        # - 그 외: self.balance -= amount
        if amount <= 0:
            print('잘못된 금액입니다.')
        elif amount > self.balance :
            print('잔액 부족합니다.')
        else :
            self.balance -= amount
    
    def get_balance(self):
        # TODO 7-4: 잔액 반환
        return self.balance


# 시나리오 테스트 (이미 작성됨 - 수정 X)
account = BankAccount("이서연", 10000)

account.deposit(5000)
print("입금 후 잔액:", account.get_balance())

account.deposit(-3000)
print("잔액:", account.get_balance())

account.withdraw(20000)
print("잔액:", account.get_balance())

account.withdraw(7000)
print("출금 후 잔액:", account.get_balance())
````

```
=== 리스트 비교 ===
list_x == list_y: True
list_x is list_y: False
list_x == list_z: True
list_x is list_z: True
id(list_x): ...
id(list_y): ...
id(list_z): ...

=== 문자열 비교 ===
str_a == str_b: True
str_a is str_b: True
...
```

````python
list_x = [1, 2, 3]
list_y = [1, 2, 3]
list_z = list_x

str_a = "hello"
str_b = "hello"

# TODO 8-1: 리스트 비교 결과 출력 (==, is, id)
print("=== 리스트 비교 ===")
print('list_x == list_y: ',list_x == list_y)
print('list_x is list_y: ',list_x is list_y)
print('list_x == list_z: ',list_x == list_z)
print('list_x is list_z: ',list_x == list_z)
id(list_x)
id(list_y)
id(list_z)

# TODO 8-2: 문자열 비교 결과 출력
print("\n=== 문자열 비교 ===")
print('str_a == str_b: ',str_a == str_b)
print('str_a is str_b: ',str_a is str_b)
````

- 천 단위 콤마: `f"₩{self.won:,}"`
- 예외 발생: `raise ValueError("...")`

````python
# 연산자 오버로딩의 핵심원리
# 내 알맹이 self.won와 상대방의 알맹이 other.won을 가지고 요리해서 결과물로 
# 새로운 Money 인스턴스를 만들어서 뱉는다가 정석이다.

'''
other은 객체이다.
print(a+b)하면, a 클래스 안에 있는 __add__(self,other)를 실행한다. 이떄 매개변수로 들어가는 녀석들의 정체
self -> a 객체 통째로. 주머니 안에 self.won = 5000을 가짐
other -> b 객체 통쨰로. 주머니 안에 other.won = 3000을 가짐

Money(self.won + other) 이렇게 작성하면
self.won 은 숫자이다. other은 money라고 적힌 가방 인스턴스이다.


2. Money(self.won + other.won) 이렇게 감싸지 않았을 때에는
그냥 8000만 반환해서 우리의 출력 양식을 따르지도 않고 (a+b) - b 같은 연속계산이 불가능하다.
8000은 그냥 int형 숫자라서 - 연산자를 썼을 때, 상대방의 가방 알맹이를 꺼내는 스킬을 쓸 수 없다.

Money(8000)으로 감싸서 반환할 때,
8000을 가지고 다시 Money(8000)을 호출한다.

새로운 인스턴스(가방)를 새로 만들어야겠다고 생각하고 메모리에 새로운 돈 객체를 찍어낸다.

이 새로 태어난 가방이 반환된다.

3. print(a+b) 했을 때.
__add__ 작동 -> 숫자 5000 + 3000 계산 -> 8000 도출
Money(8000) 실행 -> 주머니에 won = 8000을 찬 새로운 돈 객체 탄생 후 반환
print(새로운 돈 객체) 실행. 파이썬이 자동으로 이 객체의 __str__을 호출함. -> 화면에 8,000이라고 예쁘게 출력.


4. 객체명.속성명 규칙
객체지향 프로그래밍에서 속서에 접근하는 규칙의 본질
- Money 클래스의 설계도 __init__에 self.won = won이라고만 적어두었다면,
이 가방 안에는 won이라는 이름표를 가진 데이터 딱 하나만 존재한다.

즉, 속성값은 __init__에 self.xxx 로 등록해 놓은 이름만 가능하다.

---
속성의 이름은 같아야 하지만, 그 안에 들어있는 데이터는 달라도 된다.
'''


class Money:
    def __init__(self, won):
        # TODO 9-1
        self.won = won
    
    def __str__(self):
        # TODO 9-2: "₩1,000" 형식 반환
        return f"₩{self.won:,}"
    
    def __add__(self, other):
        # TODO 9-3: Money(self.won + other.won) 반환
        return Money(self.won + other.won)
    
    def __sub__(self, other):
        # TODO 9-4: 음수면 ValueError, 아니면 Money 반환

        if self.won - other.won <0:
            raise ValueError('금액은 음수가 될 수 없습니다.')
        else:
            return Money(self.won - other.won)
    
    def __eq__(self, other):
        # TODO 9-5
        return self.won == other.won

# 테스트
a = Money(5000)
b = Money(3000)

print(a + b)
print(a - b)
print(a == Money(5000))

try:
    print(b - a)
except ValueError as e:
    print("오류 발생:", e)


print(type(a))
````

## 🧪 Lab 11. `super()` 함수

### 📋 과제
Lab 10을 개선하여 각 자식 클래스가 자신의 `__init__`을 갖도록 하세요. `ElectricCar`도 추가합니다.

**Car(Vehicle)**: `__init__(name, max_speed, fuel)`  
**Bicycle(Vehicle)**: `__init__(name, max_speed, gears)`  
**ElectricCar(Vehicle)**: `__init__(name, max_speed, battery_kwh)`, `charge()` 메소드

### ✅ 예상 출력
```
아이오닉의 최고 속도는 180km/h입니다.
아이오닉: 배터리 77kWh를 충전합니다.
```

````python
class Vehicle:
    def __init__(self, name, max_speed):
        self.name = name
        self.max_speed = max_speed
    
    def info(self):
        print(f"{self.name}의 최고 속도는 {self.max_speed}km/h입니다.")


class Car(Vehicle):
    def __init__(self, name, max_speed, fuel):
        # TODO 11-1: super().__init__(...) 후 self.fuel 할당
        
        '''
        self.name = name
        self.max_speed = max_speed
        self.fuel = fuel
        '''
        super().__init__(name,max_speed)
        self.fuel = fuel
        
    def refuel(self):
        print(f"{self.name}: {self.fuel}를 주유합니다.")


class Bicycle(Vehicle):
    def __init__(self, name, max_speed, gears):
        # TODO 11-2
        '''
        self.name = name
        self.max_speed = max_speed
        '''
        super().__init__(name,max_speed)
        self.gears = gears
            
    def shift(self):
        print(f"{self.name}: {self.gears}단 기어 변속합니다.")


class ElectricCar(Vehicle):
    def __init__(self, name, max_speed, battery_kwh):
        # TODO 11-3
        '''
        self.name = name
        self.max_speed = max_speed
        '''
        super().__init__(name,max_speed)
        self.battery_kwh = battery_kwh
    
    def charge(self):
        # TODO 11-4: "이름: 배터리 ○○kWh를 충전합니다."
        return f"{self.name}: 배터리 {self.battery_kwh}kWh를 충전합니다."


# 테스트
ev = ElectricCar("아이오닉", 180, 77)
ev.info()
ev.charge()
````

## 🏁 도전 과제 (선택)

학습한 모든 개념을 활용한 미니 도서관 시스템입니다.

**요구사항**
1. `Item` 부모 클래스 (속성: `title`, `_available`, 클래스 변수 `library_name`)
2. `Book(Item)`: 추가 속성 `author`, `pages`
3. `DVD(Item)`: 추가 속성 `director`, `runtime`
4. 모든 클래스 `__str__` 정의
5. `borrow()`, `return_item()` 메소드 — `_available` 상태에 따라 분기
6. `isinstance()`로 책과 DVD를 분류하는 `categorize()` 함수

### 💡 힌트
- 부모 `__str__`을 자식에서 활용하려면 `super().__str__()` 사용
- `categorize(items)`는 `(books, dvds)` 튜플 반환

````python
class Item:
    # TODO C-1: 클래스 변수 library_name = "중앙도서관"
    library_name = "중앙도서관"
    def __init__(self, title):
        # TODO C-2: self.title, self._available = True
        self.title = title
        self._available = True
    
    def borrow(self):
        # TODO C-3: _available이 True면 False로 변경 + 대출완료 메시지
        #          False면 "이미 대출중" 메시지
    
        if self._available : # == True는 생략해도 파이썬이 알아먹는다.
            self._available = False
            print(f"'{self.title}' 대출완료")
        else:
            print(f"'{self.title}'은 이미 대출 중입니다.")
    
    def return_item(self):
        # TODO C-4: 위와 반대 동작
        if not self._available :
            self._available = True
            print(f"'{self.title}' 반납 완료")
        else:
            print(f"'{self.title}'은 이미 반납되어 대출 가능")
    
    def __str__(self):
        # TODO C-5: "[제목] - 대출가능/대출중" 형식
        return f"{self.title} - {self._available}"


class Book(Item):
    def __init__(self, title, author, pages):
        # TODO C-6
        super().__init__(title)
        self.author = author
        self.pages = pages

    def __str__(self):
        # TODO C-7: 부모 __str__ + (저자: ..., ○○p)
        return f"{super().__str__()} 저자: {self.author}, {self.pages}p"


class DVD(Item):
    def __init__(self, title, director, runtime):
        # TODO C-8
        super().__init__(title)
        self.director = director
        self.runtime = runtime
    
    def __str__(self):
        # TODO C-9
        return f"{super().__str__()} 제작자: {self.director}, 시연시간: {self.runtime}"


def categorize(items):
    # TODO C-10: isinstance를 사용해 books, dvds로 분류
    books = []
    dvds = []

    for it in items:
        if isinstance(it,Book) :
            books.append(it)
        elif isinstance(it,DVD):
            dvds.append(it)

    return books, dvds

# 테스트
items = [
    Book("파이썬 입문", "김교수", 350),
    DVD("기생충", "봉준호", 132),
    Book("객체 지향의 사실과 오해", "조영호", 224),
    DVD("올드보이", "박찬욱", 120),
]

for it in items:
    print(it)

print()
items[0].borrow()
items[0].borrow()
items[0].return_item()

print()
books, dvds = categorize(items)
print(f"책 {len(books)}권, DVD {len(dvds)}편")
````
