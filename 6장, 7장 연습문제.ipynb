## 7.4 딕셔너리의 메소드 
- key() : 딕셔너리 내 모든 키 반환
- values() : 딕셔너리 내 모든 값 반환
- items() : 딕셔너리 내의 모든 항목을 [키]:[값] 쌍으로 반환
- get(key) : 키에 대한 값을 반환한다. 키가 없으면 None을 반환한다. 
- pop(key) : 키에 대한 값을 반환하고 그 항목을 삭제한다. 키가 없으면 KeyError 예외를 발생시킨다.
- popitem() : 마지막으로 입력된 항목을 반환하고 그 항목을 삭제한다.
- clear() : 딕셔너리 내의 모든 항목을 삭제한다.

```python
person = {'이름':'홍길동','나이':26,'몸무게':82}
print(person.keys())
print(person.values())
print(person.items()) # 개별 항목을 튜플 형태로 반환

# person['취미'] = 키에러 발생시킨다.
print(person.get('취미'))
print(person.get('이름'))

# None 키워드는 아무 값이 없을 때 사용하는 키워드이다.
# 0이나 False 같은 값을 이용하여 '아무 값 없음'을 표현할 수 있을 것 같지만,
# 이들은 엄연한 객체이며, 변수가 '아무런 값도 가지지 않음'을 표현하는 유일한 방법은 None 을 할당하는 방법이다.

print(person.popitem())
print(person)
person.pop('나이')
print(person)

person2 = {'이름':'홍길동','나이':26,'몸무게':82}
print(person2.clear())
```

#### 딕셔너리 항목을 하나하나 열거하기 위해서는 for ~ in 문을 사용한다.

## 7.8 함수 반환문과 튜플 활용

튜플을 이용하여 여러 개의 값을 반환하는 법을 배워보자.

원의 면적과 둘레 값인 area circum을 각각의 수식을 이용하여 계산하는 것을 볼 수 있다.

area, circum 쌓으로 그 결과를 반환하면 튜플 값이 반환되는 것도 알 수 있다.

패킹, 언패킹을 이용하여 튜플 값에 접근하여 처리할 수 있다.

함수 내부에서는 area, circum 값을 쉼표로 구분하여 패킹하고 함수 외부에서는 a,c 변수가 두 값을 언패킹하는 방법으로 값을 전달받는다.

```python
def area_and_circum(radius):
    area = 3.14 * radius **2
    circum = 2*3.14*radius
    return area,circum
r = 4
a, c = area_and_circum(r)
print(f'반지름 {r}인 원의 면적과 둘레 : {a}, {c}')
```

```python
def product_set(set1, set2):
    res = set()
    for i in set1:
        for j in set2:
            res.add((i,j))
    return res

def exp(input_set,exponent):
    res = input_set
    for _ in range(exponent-1):
        res = product_set(res,input_set)
    return res

A = {1,3}
A3 = exp(A,3)
print(A3)
```

# 7.12 zip() 함수를 이용한 집적화와 슬라이싱
파이선에서 리스트, 딕셔너리, 집합, 튜플과 같은 자료형을 반복가능자료형이라고 한다. iterable

이러한 반복 가능 자료형 여러 개를 넘겨주면, 이들을 합쳐서 튜플 반복자(iterator)를 반환하는 함수가 zip() 함수이다.

zip() 함수는 다음과 같이 여러 개의 반복 가능 자료형을 받을 수 있는데, 이를 집적화 (aggregation)라고 한다.

zip(*iterables)

---

인자가 전달되지 않으면, zip() 함수는 공백 반복자를 반환한다. 하나의 반복 가능 자료형 인자가 들어올 경우 zip() 함수는 튜플형 반복 가능 자료형을 반환한다.

이 튜플의 각 원소는 인자로 넘어온 데이터에서 얻어진 값 하나씩이 전달된다. 다수의 반복 가능 자료형 n개가 인자로 들어오면 n개의 요소를 가진 튜플이 반복적으로 나타난다.

이때 i번째 튜플은 인자로 넘어온 각 반복 가능 자료형 인자의 i번쨰 원소들로 구성된다.



```python
empty_iterator = zip()
result = set(empty_iterator)
print(result)

'''
빈 공백이 인자로 넘겨진 경우
zip() 함수는 공백 반복자를 생성하고 이 공백 반복자를 이용하여 
원소를 얻어 집합을 만든 결과를 result에 할당하였다.

이 result는 공집합이다.
'''

a= [10,20,30]
b = 'ten','twenty','thirty'
print(list(zip(a,b)))

'''
zip() 함수가 반환하는 것은 zip 객체인데 이는 반복 가능 객체이므로
list() 함수를 사용하여 리스트 객체로 만드는 것이 가능하다.

이 출력은 위의 결과와 같이 a의 각 요소와 b의 각 요소로 이루어진 튜플이며

'''

# zip 함수에 하나의 반복 가능 자료형 인자를 전달한 경우

lst = ['a','b','c','d']
my_iterator = zip(lst)
result = set(my_iterator)
print(result)

cty_list = ['서울','부산','광주','대전']
pop_tuple = 9765,3467,1450
area_list = [605.2, 769.9, 480, 496, 833]
cty_info = zip(cty_list,pop_tuple,area_list)
# list(cty_info)
# print(list(cty_info))

a,b,c = zip(*cty_info)
print(a)
print(b)
print(c)
```
