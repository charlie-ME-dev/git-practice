# 8.1 모듈과 import 문법
모듈 : 파이썬 함수나 변수 또는 클래스들을 모아 놓은 스크립트 파일을 말한다.

개발된 모듈은 표준 라이브러리라는 이름으로 파이썬 설치 파일과 함께 제공되고 있다.

미리 만들어진 모듈을 가져올 때에는 'import'와 함께 모듈 이름을 써주면 되고 사용할 때에는 모듈 이름에 점을 찍은 후 모듈 아느이 구성요소를 호출 해 주면 된다.

#### 파이썬 표준 라이브러리
- 문자열과 텍스트 처리를 위한 모듈
- 이진 데이터 처리
- 날짜
- 시간 
- 배열 등의 자료형 처리를 위한 모듈
- 수치 연산과 수학 함수 모듈
- 파일과 디렉토리 접근
- 유닉스 시스템의 데이터베이스 접근을 위한 모듈
- 데이터 압축
- 그래픽 모듈 등 100가지 이상의 표준 라이브러리가 있다.

몇 개의 편리한 모듈과 그 이용 방법을 알아보자.

# 8.2 날짜와 시간 모듈 datetime
import datetime

datetime 모듈에 있는 datetime 클래스는 날짜와 시간에 관련된 여러가지 기능을 제공한다.

이 클래스의 함수에 접근하기 위해서는 datetime.datetime과 같이 클래스 이름 다음에 마침표 연산자를 써주면 된다.

datetime.datetime 에서 앞의 것은 모듈이고 뒤의 것은 클래스의 이름이 된다.

이 클래스의 now() 메소드를 사용하면, 년, 월, 일, 시, 분, 초, 밀리초 단위의 시간까지 나타난다.

문자열과 마찬가지로 마침표 연산자를 사용하여 클래스의 메소드를 호출한다.

---
date 클래스는 날짜에 관련된 정보를 포함하고 있다.

이를 이용하면 **오늘의 날짜**를 간단하게 살펴볼 수 있다.

다음과 같은 방법으로 datetime 의 date 클래스에 있는 todat()라는 메소드를 호출하자.

현재 date 클래스의 todat() 메소드는 현재 날짜를 todat에 반환하는데 이 today를 pinrt() gkatnfh cnffurgo qhaus 2025-01-02와 같은 datetime.date 클래스가 가진 년, 월, 일, 을 출력해준다.

년도만 보고자 할 때에는 todat.year라는 속성값을 사용한다.

```python
import datetime
print(datetime.datetime.now())

today = datetime.date.today()
print(today)
print(today.year)
print(today.month)
print(today.day)

```
