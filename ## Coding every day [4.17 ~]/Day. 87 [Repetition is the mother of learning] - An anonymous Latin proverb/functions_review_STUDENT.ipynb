## 1. 함수 기초 — `def` 와 `return` / Function basics — `def` and `return`

함수는 `def 이름(매개변수):` 로 정의하고, `return` 으로 결과 값을 돌려줍니다. `return` 이 없으면 함수는 `None` 을 반환해요.

A function is defined with `def name(params):` and hands back a result with `return`. Without `return`, a function returns `None`.

| 구성요소 / Part | 의미 / Meaning |
|---|---|
| `def` | 함수 정의 시작 / starts a function definition |
| `매개변수 / parameter` | 함수가 받는 입력값 / the input a function takes |
| `return` | 결과 값을 돌려줌 / sends a value back |

**🔎 예제 / Worked example**

```python
# 정사각형 넓이를 반환하는 간단한 함수
# A simple function returning a square's area
def square_area(side):
    return side * side

print(square_area(5))   # 25
```

## 5. 여러 값 반환 / Returning multiple values

`return a, b` 처럼 쉼표로 여러 값을 반환하면 **튜플**로 묶여 나옵니다. 받을 때는 `x, y = ...` 로 **언패킹** 할 수 있어요.

`return a, b` returns several values bundled as a **tuple**. You can **unpack** them with `x, y = ...`.

**🔎 예제 / Worked example**

```python
# 최솟값과 최댓값을 함께 반환 / returns min and max together
def min_max(a, b):
    return min(a, b), max(a, b)

low, high = min_max(8, 3)
print(low, high)   # 3 8
```

## 6. `*args` — 가변 위치 인자 / `*args` — variable positional arguments

`*args` 를 쓰면 **개수가 정해지지 않은** 인자들을 하나의 튜플로 받을 수 있습니다. 인자가 0개여도 동작해요.

`*args` lets a function accept **any number** of arguments as a single tuple. It works even with zero arguments.

**🔎 예제 / Worked example**

```python
# 임의 개수의 숫자를 모두 더함 / sums any number of values
def add_all(*nums):
    total = 0
    for n in nums:
        total += n
    return total

print(add_all(1, 2, 3, 4))   # 10
print(add_all())             # 0
```

## 7. `**kwargs` — 가변 키워드 인자 / `**kwargs` — variable keyword arguments

`**kwargs` 는 `이름=값` 형태의 인자들을 **딕셔너리**로 받습니다. 동적으로 항목을 추가할 때 유용해요.

`**kwargs` collects `name=value` arguments into a **dictionary**. Handy for building records dynamically.

**🔎 예제 / Worked example**

```python
# 받은 키워드 인자를 딕셔너리로 출력 / prints received keywords as a dict
def show_options(**options):
    return options

print(show_options(color="red", size="L"))   # {'color': 'red', 'size': 'L'}
```

## 8. 스코프 — 지역 변수와 전역 변수 / Scope — local vs global

함수 **안에서** 만든 변수는 **지역(local)** 변수로, 함수 밖에서는 보이지 않습니다. 함수 **밖**의 **전역(global)** 변수는 함수 안에서 **읽을 수** 있어요.

A variable made **inside** a function is **local** and invisible outside. A **global** variable (defined outside) can be **read** inside a function.

**🔎 예제 / Worked example**

```python
# 전역 변수를 함수 안에서 읽기 / reading a global inside a function
TAX_RATE = 0.1

def tax_on(price):
    return price * TAX_RATE   # 전역 TAX_RATE 를 읽음 / reads global

print(tax_on(1000))   # 100.0
```

## 9. 중첩 함수 (헬퍼 함수) / Nested functions (helper functions)

함수 **안에** 또 다른 함수를 정의할 수 있습니다. 이런 **내부(헬퍼) 함수**는 바깥 함수 안에서만 쓰이며, 복잡한 계산을 깔끔하게 나눌 때 유용해요.

You can define a function **inside** another. This **inner (helper) function** is used only within the outer one — great for splitting up a calculation cleanly.

**🔎 예제 / Worked example**

```python
# 헬퍼 함수로 계산을 분리 / split work into a helper
def outer(x):
    def double(n):       # 내부 헬퍼 / inner helper
        return n * 2
    return double(x) + 1

print(outer(5))   # 11
```

**✅ 연습 12 / Practice 12 — `exchange_with_fee`**

Return the **net KRW after fee** for `usd`. Defaults: `rate` **1300**, `fee_rate` **0.01**. (won = usd × rate; net = won − won×fee_rate)

```python
def exchange_with_fee(usd, rate=1300, fee_rate=0.01):
    won = usd*rate          # TODO 1: usd * rate
    fee = won*fee_rate          # TODO 2: won * fee_rate
    return won - fee         # TODO 3: won - fee
```

## 13. 종합 ④ — 캡스톤: `*args` + `**kwargs` 함께 / Applied ④ — Capstone: `*args` + `**kwargs` together

마지막! 일반 매개변수, `*args`, `**kwargs` 를 **한 함수에** 모두 사용합니다. 순서는 항상 `(일반, *args, **kwargs)` 입니다.

Last one! Use a regular parameter, `*args`, and `**kwargs` **in one function**. The order is always `(regular, *args, **kwargs)`.

**🔎 예제 / Worked example**

```python
# 셋을 모두 사용 / uses all three
def make_log(level, *messages, **fields):
    return {"level": level, "messages": list(messages), "extra": fields}

print(make_log("INFO", "start", "ok", user="mina"))
# {'level': 'INFO', 'messages': ['start', 'ok'], 'extra': {'user': 'mina'}}
```

**✅ 연습 13 / Practice 13 — `build_record`**
거래 기록을 만드세요. `account`(계좌명), 이어서 **개수 제한 없는 거래 금액들** `*transactions`, 그리고 추가 정보 `**meta` 를 받습니다.
반환 딕셔너리: `{"account": 계좌명, "transactions": [금액들], "total": 합계, ...meta}`
Build a transaction record. Take `account`, then **any number of** `*transactions`, plus `**meta`.
Return a dict: `{"account": account, "transactions": [amounts], "total": sum, ...meta}`

```python
def build_record(account, *transactions, **meta):
    record = {
        "account": account,
        "transactions": list(transactions),   # TODO 1: transactions 를 리스트로 / transactions as a list
        "total": sum(transactions),          # TODO 2: 거래들의 합계 / sum of transactions
    }
    # TODO 3: meta 의 항목들을 record 에 합치기 / merge meta into record
    record.update(meta)
    return record
```

```python
# 테스트 / Test
r = build_record("Mina", 1000, 2000, status="ok")
expected = {"account": "Mina", "transactions": [1000, 2000], "total": 3000, "status": "ok"}
if r == expected and build_record("Jun") == {"account": "Jun", "transactions": [], "total": 0}:
    print("✅ 통과 / Passed:", r)
else:
    print("❌ 다시 시도 / Try again:", r)
```

**정답 11 / Solution 11 — `monthly_summary`**

```python
def monthly_summary(*amounts):
    count = len(amounts)
    total = sum(amounts)
    if count == 0:
        average = 0
    else:
        average = total / count
    return count, total, average

print(monthly_summary(1000, 2000, 3000))   # (3, 6000, 2000.0)
```
