```python
'''
0. 준비
1. 파일 읽기
- f.read() = 파일 전체를 하나의 문자열
- f.readlines() = 줄들의 리스트
- for line in f = 한줄 씩 반복 (줄 단위로 처리할 때. = 가장 흔하다.)
- - [int(line.stript()) for line in f if line.strip()]
'''

# f.read() 파일 전체를 하나의 문자열. 파일이 작고 전체가 필요할 때
# f.readline() 줄들의 리스트. 줄 목록이 필요할 때
# for line in f 한 줄씩 반복. 줄 단위로 처리할 때

with open('diary.txt','w',encoding='utf-8') as f: # with 문법의 경우, close가 불필요하다.
    f.write('오늘은 Python 파일 입출력을 배웠다.\n')
    f.write('생각보다 어렵지 않았다!\n')
    f.write('내일은 CSV도 배운다.\n')


file_pra = open('diary.txt','r',encoding='utf-8')
content = file_pra.read()
file_pra.close()

print(content)
print(type(content))

file_pra = open('diary.txt','r',encoding='utf-8')

for line in file_pra: # 한줄씩반복을 의미.
    print('-',line.strip())

file_pra.close()

file_pra = open('diary.txt','r',encoding='utf-8')
lines = file_pra.readlines()
file_pra.close()

print(lines)
print(len(lines))

print('-'*30)

# [실습1] Wonder Exchange 환율 로그 읽기 / Read the rate log

f = open('wonder_exchange_rates.txt','r',encoding='utf-8')
rates = [int(new_line.strip()) for new_line in f if new_line.strip()]
avg_rates = sum(rates) / len(rates)
print(rates)
print(f'{avg_rates:.2f}')
f.close()

# 조건문은 값이 비어있으면 false, 뭔가 들어있으면 true로 판단한다.
# 그래서 if new_line.strip()
# 이렇게 작성하면 new_line.strip()안에 뭔가 들어있어야 해당 문이 시작된다.
# 리스트 컴프리헨션 더 공부하기.

print('-'*30) # 파일쓰기 / Writing Files

'''
2. 파일쓰기
- write() = 한줄 씩 쓰기
- writelines() - 리스트를 한 번에
이것은 줄바꿈을 자동으로 넣어주지 않는다. 직접 \n을 포함해야 한다.
'''


with open('cafe_orders.txt','w',encoding='utf-8') as f:
    f.write('아메리카노\n')
    f.write('카페라떼\n')

with open('cafe_orders.txt','r',encoding='utf-8') as f:
    print(f.read())
    print(f.readlines()) # 파이썬은 한번 읽으면 다시 읽지 않아서 이거 실행하면 아무것도 안뜬다. [] 이렇게 나옴.
    print(len(f.readlines())) # 마찬가지로 파이썬은 한 번 읽으면 다시 읽지 않아서 이것도 0 나온다.

menu = ['바닐라라떼\n', '콜드브루\n', '에스프레소\n']

with open('cafe_orders.txt','w',encoding='utf-8') as f: # 'w' 모드로 열면, 백지부터 다시쓴다!
    f.writelines(menu)
# 내용을 이어가고 싶으면 'a' 모드로 들어가야 한다.

ff = open('cafe_orders.txt','r',encoding='utf-8')

a = ff.read()
print(a)

ff.close()

# 우리가 f 변수에 파일 열었다가 닫으면, 파일은 닫혀있는 좀비상태이다.
# 변수 f는 책을 쥐고 있는 손이라고 생각하면 좋다.
# close 하거나 with 블록 빠져나오면 '책을 덮는 행위'일 뿐, f 자체가 사라지는 게 아니다.
# 파이썬에서 변수 이름 f 는 그저 데이터를 가리키는 포스트잇일 뿐이다.
# 파이썬에서는 가비지 컬렉터가 있어서 변수 이름표를 이리저리 마음껏 옮겨 붙이며 코딩할 수 있다.
# 즉, a = 2 a = 3 이렇게 하면 2는 사라지는데, 이것을 파이썬 내부에서 메모리 먹지 않게끔 한다.

# ---

# 실습2 Wonder Pay 영수증 쓰기 / Write a receipt
items = [('아메리카노',4500),('샌드위치',6000)]

with open('Wonder_pay_receipt.txt','w',encoding='utf-8') as f:
    f.write('=== Wonder Pay 영수증 ===')
    for name, price in items: 
        # for문을 이용해서 튜플 언패킹
        f.write(f"{name} : {price}원\n")
    f.write(f'합계: {sum(price for _, price in items)}원\n')

with open('Wonder_pay_receipt.txt','r',encoding='utf-8') as f:
    print(f.read())

print('-'*30)

'''
3. 파일에 추가하기
'a' 모드는 기존 내용을 지우지 않고 파일끝에 이어서 쓴다.
파일이 없으면 'a'와 'w' 모두 새로 만들어준다. 차이는 기존 내용 보존 여부이다.
'''
# 예제 3-1 일기 한 줄 추가

with open('diary.txt','r',encoding='utf-8') as f:
    print('추가 전 / before', len(f.readlines()), '줄')

with open('diary.txt','a',encoding='utf-8') as f:
    f.write('파일 추가(append) 모드도 배웠다.\n')

with open('diary.txt','r',encoding='utf-8') as f:
    print('추가 후 / after:', len(f.readlines()),'줄')

with open('diary.txt','r',encoding='utf-8') as f:
    print('---\n' + f.read())
```
