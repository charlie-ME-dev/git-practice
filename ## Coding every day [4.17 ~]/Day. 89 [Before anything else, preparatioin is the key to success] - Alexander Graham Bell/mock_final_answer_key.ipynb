## Part 2 · 버그 찾기 정답 / Find-the-Bug Answers
채점: 버그를 정확히 설명 **2점** + 올바르게 수정 **3점** = 문항당 5점.
*Grading: correct identification **2 pts** + correct fix **3 pts** = 5 pts each.*

**B1** · 문자열/리스트
- 🐞 버그 / Bug: `name = name.title()`는 반복 변수만 다시 묶을 뿐 리스트를 바꾸지 않음 (문자열은 불변).
  *`name = name.title()` only rebinds the loop variable; it doesn't change the list (strings are immutable).*
- ✅ 고친 출력 / Fixed output: `['Kim Minji', 'Lee Junho']`

```python
def format_names(raw):
    return [name.title() for name in raw.split(",")]
print(format_names("kim minji,lee junho"))
```

**B2** · 딕셔너리
- 🐞 버그 / Bug: 처음 보는 키에 `counts[g] += 1`을 하면 키가 없어 KeyError 발생.
  *`counts[g] += 1` on a brand-new key raises KeyError because the key doesn't exist yet.*
- ✅ 고친 출력 / Fixed output: `{'fiction': 2, 'poetry': 1}`

```python
def count_genres(genres):
    counts = {}
    for g in genres:
        counts[g] = counts.get(g, 0) + 1
    return counts
print(count_genres(["fiction","fiction","poetry"]))
```

**B3** · 세트/튜플
- 🐞 버그 / Bug: 리스트는 해시 불가능해서 set에 넣으면 TypeError. 변하지 않는 튜플로 바꿔야 함.
  *Lists are unhashable, so adding one to a set raises TypeError; convert to an immutable tuple.*
- ✅ 고친 출력 / Fixed output: `2`

```python
def unique_visitors(log):
    seen = set()
    for entry in log:
        seen.add(tuple(entry))
    return len(seen)
print(unique_visitors([[1,2],[1,2],[3]]))
```

**B4** · 제어 흐름
- 🐞 버그 / Bug: `range(len(prices) - 1)`은 마지막 요소를 빠뜨림 (off-by-one).
  *`range(len(prices) - 1)` skips the last element (off-by-one).*
- ✅ 고친 출력 / Fixed output: `6500`

```python
def total_spent(prices):
    total = 0
    for p in prices:
        total += p
    return total
print(total_spent([3000,2000,1500]))
```

**B5** · 함수
- 🐞 버그 / Bug: 가변 기본 인자 `bag=[]`는 모든 호출이 같은 리스트를 공유 → 값이 누적됨.
  *Mutable default `bag=[]` is shared across all calls, so items accumulate.*
- ✅ 고친 출력 / Fixed output: `['pillow']
['blanket']`

```python
def add_item(item, bag=None):
    if bag is None:
        bag = []
    bag.append(item)
    return bag
print(add_item("pillow"))
print(add_item("blanket"))
```

**B6** · 함수/제어 흐름
- 🐞 버그 / Bug: `else: return False`가 첫 요소에서 바로 반환 → 나머지를 검사하지 못함. `return False`는 반복문 밖에 있어야 함.
  *`else: return False` returns on the first item, so the rest is never checked; `return False` belongs after the loop.*
- ✅ 고친 출력 / Fixed output: `True`

```python
def has_overdue(days_list):
    for days in days_list:
        if days > 14:
            return True
    return False
print(has_overdue([10,20,5]))
```

**B7** · OOP
- 🐞 버그 / Bug: `balance = balance`는 지역 변수에만 대입 → `self.balance`가 만들어지지 않아 AttributeError.
  *`balance = balance` assigns only a local; `self.balance` is never created → AttributeError.*
- ✅ 고친 출력 / Fixed output: `7000`

```python
class MealCard:
    def __init__(self, balance):
        self.balance = balance
    def pay(self, amount):
        self.balance -= amount
        return self.balance
print(MealCard(10000).pay(3000))
```

**B8** · 예외 처리
- 🐞 버그 / Bug: 0으로 나누면 ZeroDivisionError인데 ValueError를 잡으려 해서 처리되지 않고 프로그램이 멈춤.
  *Dividing by zero raises ZeroDivisionError, but the code catches ValueError, so it isn't handled and crashes.*
- ✅ 고친 출력 / Fixed output: `0`

```python
def divide_fare(total, riders):
    try:
        result = total / riders
    except ZeroDivisionError:
        result = 0
    return result
print(divide_fare(9000,0))
```
