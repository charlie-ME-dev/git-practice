# Problem 1. 숫자 줄이기 게임
### 숫자 하나로 시작하는 게임입니다. 숫자가 짝수이면 2로 나누고, 홀수이면 3을 곱한 뒤 1을 더합니다. 이 과정을 숫자가 1이 될 때까지 반복했을 때, 몇 번 만에 1에 도달하는지(단계 수)를 반환하세요. 1보다 작은 수가 들어오면 ValueError를 발생시킵니다.

```python
def collatz_steps(number):
    if number < 1:
        raise ValueError
        
    i = 0
    
    while(1):
        if number == 1:
            break        
        elif number % 2 == 0:
            i += 1
            number = number /2
        else :
            i += 1
            number = number * 3 + 1
    return i

print(collatz_steps(1)) # 0
print(collatz_steps(16)) # 4
print(collatz_steps(12)) # 9
try:
    print(collatz_steps(0)) # ValuError 발생
except ValueError:
    print("입력값은 1 이상의 자연수여야 합니다.")
    
```

# Problem 2. 다트 점수 계산기
### 다트 과녁의 중심은 좌표 (0, 0)입니다. 다트가 꽂힌 위치 (x, y)를 받아 점수를 계산하세요. 중심에서의 거리에 따라 점수가 정해집니다.
1.	거리가 1 이하이면 10점
2.	거리가 5 이하이면 5점
3.	거리가 10 이하이면 1점
4.	그보다 멀면 0점


```python
def dart_score(x, y):
    dart_distance = (x**2 + y**2)**0.5
    score = 0
    if dart_distance <= 1:
        score = 10
    elif dart_distance <= 5:
        score = 5
    elif dart_distance <= 10:
        score = 1
    else:
        score = 0
        
    return score 
    
print(dart_score(0,0)) # 10
print(dart_score(-5,0)) # 5
print(dart_score(0,10)) # 1
print(dart_score(-9,9)) # 12.7
```

# Problem 3. 약어 만들기
### 긴 이름이나 구절을 약어(머리글자)로 줄이는 기능을 만듭니다. 각 단어의 첫 글자를 대문자로 모아 이어 붙이세요. 단어는 공백과 하이픈(-)으로 구분됩니다.

```python
def make_acronym(phrase):
    phrase = phrase.title()
    phrase = phrase.replace("-"," ")
    phrase = phrase.split()
    result = []
    for phrase_str in phrase:
        for n in phrase_str:
            if n == n.upper():
                result.append(n)
    result = "".join(result)
    return result

print(make_acronym("Portable Network Graphics")) # PNG
print(make_acronym("Ruby on Rails")) # ROR
print(make_acronym("Complementary metal-oxide semiconductor")) # CMOS
print(make_acronym("Hyper-Text Markup Language")) # HTML
```
